```python
class shape:
    def calculate_area(self):
        pass
    
class circle(shape):
    def __init__(self,radius) -> None:
        self.radius = radius
    
    def calculate_area(self):
        return 3.14* (self.radius)
    
class square(shape):
    def __init__(self,side) -> None:
        self.side = side 
    def calculate_area(self):
        return self.side*self.side
    
x = circle(5)        
y = square(4)
print("Area:", x.calculate_area())        
print("Area:", y.calculate_area())       
```

```python
class MOdel:
    def pridect(self,x):
        raise
    
class linearRegression(MOdel):
    def pridect(self, x):
        return "linear Regression prediction"
    
class DicisionTree(MOdel):
    def pridect(self,x):
        return "Dicision Tree prediction " 
    
models = [linearRegression(),DicisionTree()] 
for model in models:
    print(model.pridect(None))         
```
