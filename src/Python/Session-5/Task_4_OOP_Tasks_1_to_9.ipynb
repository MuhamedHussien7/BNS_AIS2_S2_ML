## Task 1

```python
class Task1:
    """
    Task1 class converts a tuple to a list, adds an item, and converts it back to a tuple.
    
    Attributes:
        my_tuple (tuple): The original tuple.
        item_to_add (any): The item to be added to the tuple.
    """
    def __init__(self, my_tuple, item_to_add):
        self.my_tuple = my_tuple
        self.item_to_add = item_to_add

    def add_item(self):
        """
        Adds the item to the tuple by converting it to a list first.
        
        Returns:
            tuple: The new tuple containing the added item.
        """
        temp_list = list(self.my_tuple)
        temp_list.append(self.item_to_add)
        return tuple(temp_list)

t1 = Task1((1, 2, 3), 4)
print(t1.add_item())
```

## Task 2

```python
class Task2:
    """
    Task2 class calculates the total sum of all items in a list.
    
    Attributes:
        numbers (list): List of numbers to be summed.
    """
    def __init__(self, numbers):
        self.numbers = numbers

    def sum_items(self):
        """
        Calculates the sum of list items using a loop.
        
        Returns:
            int/float: Total sum of the numbers in the list.
        """
        total = 0
        for num in self.numbers:
            total += num
        return total

t2 = Task2([5, 10, 15, 20])
print(t2.sum_items())
```

## Task 3

```python
class Task3:
    """
    Task3 class multiplies all items in a list.
    
    Attributes:
        numbers (list): List of numbers to be multiplied.
    """
    def __init__(self, numbers):
        self.numbers = numbers

    def multiply_items(self):
        """
        Multiplies all elements in the list together.
        
        Returns:
            int/float: Product of all numbers in the list.
        """
        result = 1
        for num in self.numbers:
            result *= num
        return result

t3 = Task3([2, 3, 4])
print(t3.multiply_items())
```

## Task 4

```python
class Task4:
    """
    Task4 class finds the smallest number in a list without using built-in min().
    
    Attributes:
        numbers (list): List of numbers to evaluate.
    """
    def __init__(self, numbers):
        self.numbers = numbers

    def find_smallest(self):
        """
        Finds the smallest value manually.
        
        Returns:
            int/float: Smallest number found in the list.
        """
        smallest = self.numbers[0]
        for num in self.numbers:
            if num < smallest:
                smallest = num
        return smallest

t4 = Task4([10, 3, 25, 1, 7])
print(t4.find_smallest())
```

## Task 5

```python
class Task5:
    """
    Task5 class finds the largest number in a list without using built-in max().
    
    Attributes:
        numbers (list): List of numbers to evaluate.
    """
    def __init__(self, numbers):
        self.numbers = numbers

    def find_largest(self):
        """
        Finds the largest value manually.
        
        Returns:
            int/float: Largest number found in the list.
        """
        largest = self.numbers[0]
        for num in self.numbers:
            if num > largest:
                largest = num
        return largest

t5 = Task5([10, 3, 25, 1, 7])
print(t5.find_largest())
```

## Task 6

```python
class Task6:
    """
    Task6 class counts the number of strings with a length greater than or equal to 2.
    
    Attributes:
        words (list): List of string elements.
    """
    def __init__(self, words):
        self.words = words

    def count_strings(self):
        """
        Counts strings meeting the length condition.
        
        Returns:
            int: Count of matching strings.
        """
        count = 0
        for word in self.words:
            if len(word) >= 2:
                count += 1
        return count

t6 = Task6(["hi", "a", "hello", "no", "on"])
print(t6.count_strings())
```

## Task 7

```python
class Task7:
    """
    Task7 class creates a shallow copy of a given list.
    
    Attributes:
        original (list): The original list to copy.
    """
    def __init__(self, original):
        self.original = original

    def clone_list(self):
        """
        Copies the list.
        
        Returns:
            list: Cloned copy of the original list.
        """
        return self.original.copy()

t7 = Task7([1, 2, 3, 4])
print(t7.clone_list())
```

## Task 8

```python
class Task8:
    """
    Task8 class handles removing an item from a set.
    
    Attributes:
        my_set (set): The target set to modify.
        item (any): The item to remove from the set.
    """
    def __init__(self, my_set, item):
        self.my_set = my_set
        self.item = item

    def remove_item(self):
        """
        Removes specified item from the set.
        
        Returns:
            set: Updated set after removal.
        """
        self.my_set.remove(self.item)
        return self.my_set

t8 = Task8({1, 2, 3, 4, 5}, 3)
print(t8.remove_item())
```

## Task 9

```python
class Task9:
    """
    Task9 class checks if one set is a subset of another set.
    
    Attributes:
        set1 (set): The potential subset.
        set2 (set): The main set to compare against.
    """
    def __init__(self, set1, set2):
        self.set1 = set1
        self.set2 = set2

    def check_subset(self):
        """
        Determines whether set1 is subset of set2.
        
        Returns:
            bool: True if set1 is a subset, otherwise False.
        """
        return self.set1.issubset(self.set2)

t9 = Task9({1, 2}, {1, 2, 3, 4})
print(t9.check_subset())
```
