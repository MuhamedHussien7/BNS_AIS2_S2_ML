```python
text = "###!!@mocleW EPGTQ!!!6789"
core_part = text.lstrip("###!!@").rstrip("!6789")
word1, word2 = core_part.split()
word1_reversed = word1[::-1]
new_word1 = word1_reversed.replace("Welcom","Welcome")
new_word2 = word2.replace ( "A" , "" ) .replace ( "E" , "" ) .replace ( "I" , "" ) .replace ( "O" , "" ) .replace ( "U" , "" ) .replace ( "a" , "" ) .replace ( "e" , "" ) .replace ( "i" , "" ) .replace ( "o" , "" ) .replace ( "u" , "" )
result = new_word1 + " " + new_word2
print(result)
```
