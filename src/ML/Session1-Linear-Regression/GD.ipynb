```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Data
x = np.array([1, 2, 3, 4])
y = np.array([2, 2.8, 3.6, 4.5])
n = len(x)
#initialize parameters function
w = 0  
b = 0
# print(w,b)
# 0.63*x + 0.9 = y`
```

```python
#Hyperparameters 
learning_rate = 0.01
epoch = 100
cost_values = []
print(f"Initial w: {w}" 
      f"Initial b: {b}"
      f"Learning rate: {learning_rate}"
      f"Epoch: {epoch}")
print("training started......")
```

```python
# Gradient Descent Algorithm
for i in range(epoch):
    #prediction
    y_hat = w * x + b
    #error
    error = y_hat - y
    #Gradient calculation
    Dw = np.mean(error * x)
    Db = np.mean(error)
    #update parameters
    w = w - learning_rate * Dw
    b = b - learning_rate * Db
    y_hat = w * x + b
    # cost function
    cost = 0.5*np.mean(y_hat - y)**2
    cost_values.append(cost)

if (i + 1) % 10 == 0 or i == 0:
    print(f"Epoch:{i+1}  "
         f"Cost= {cost} , "
         f"w= {w} , "
         f"b= {b}  ")
    
```

```python
plt.figure(figsize=(8, 5))
plt.subplot(1, 2, 1)
plt.scatter(x, y)
plt.plot(x, y_hat, color='r')
plt.title("Linear Regression")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(cost_values)
plt.title("Cost Vs Epochs")
plt.xlabel("Epochs")
plt.ylabel("Cost")
plt.grid()
```
