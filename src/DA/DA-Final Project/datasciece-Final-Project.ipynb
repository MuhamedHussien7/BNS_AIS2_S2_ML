

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with mean for numeric and mode for categorical

```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].mean())
```

```python
print(df.isnull().sum())
```

### **And with a proper category for other (member gender)**

```python
print([df['member_gender'].value_counts(dropna=False)])
# print(gbf.member_gender.head(15))
print(df['member_gender'].head(15))
```

```python
df['member_gender'] = df['member_gender'].fillna('other')
```

```python
print(df['member_gender'].head(15))
```

```python
print(df['member_gender'].isnull().sum())
```

```python
def get_average_age(dataframe):
    avg_age = dataframe['member_age'].mean()
    return f"The average users age is: {avg_age:.2f} years"

print(get_average_age(df))
```

# <a id="15">  15 ) What is the  Correlation between data_stat variables in the dataset?</a>


```python
import matplotlib.pyplot as plt
import seaborn as sns

if 'member_age' not in df.columns and 'member_birth_year' in df.columns:
    df['member_age'] = 2018 - df['member_birth_year']

plt.figure(figsize=(8, 6), dpi=100)

stat_vars = df[['duration_sec', 'duration_min', 'member_birth_year', 'member_age', 'bike_id']]
corr_matrix = stat_vars.corr()

sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, cbar=True, linewidths=0.5)

plt.title('Correlation Matrix of Statistical Variables', fontsize=12, fontweight='bold', pad=15)

plt.show()
```
